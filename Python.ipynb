# Boolean Operators

Boolean operators are used in 'for' and 'while' loops (we'll see them later) as well as in other functions to apply logical conditions.

| Operator | Description |
|--------|----------|
| a == b | a equal to b? |
| a != b | a not equal to b? |
| a > b | a greater than b? |
| a >= b | a greater than or equal to b? |
| a < b | a less than b? |
| a <= b | a less than or equal to b? |
| a is b | a the same object as b? |
| a not b | only a not the same object as b? |
| a or b | a the same object as b? |

# Arithmetic Operators
Arithmetic operators are used in 'for' and 'while' loops as well, but also for data manipulation and mathematical calculations (who would've guessed?)

| Operator | Description |
|--------|----------|
| + | addition |
| - | subtraction |
| / | division |
| * | multiplication |
| a % b | called modulus, divides a by b and spits out the remainder |
| a += 1 | equal to a plus 1,  equal to a+1 |
| a -= 1 | equal to a minus 1, equal to a-1 |



````python
# Arithmetic Operators
a = 6
b = 3

# Addition and multiplication
result1 = (a + b) * 2
print("Result of (a + b) * 2:", result1)

# Subtraction and division
result2 = (a - b) / 2
print("Result of (a - b) / 2:", result2)

# Modulo
result3 = (a % b)
print("Result of (a % b) :", result3)

# exponentiation
result4 = (a ** b)
print("Result of (a ** b) :", result4)

# Boolean Operators
x = True
y = False

# AND and NOT
result5 = x and not y
print("Result of x and not y:", result5)

# OR and equality comparison
result6 = x or (a == b)
print("Result of x or (a == b):", result6)

# Greater than and less than or equal to
result7 = (a > b) and (b <= 5)
print("Result of (a > b) and (b <= 5):", result7)

# NOT and inequality comparison
result8 = not (a != 10)
print("Result of not (a != 10):", result8)
````

# Loops
Loops allow us to execute blocks of code multiple times, for example if you want to read through every row of a dataset, or test different values within a function

## For Loops
For loops allow us to execute a block of code a specific number of times.

It iterates through an iterable variable (list, tuple, range, string), executing the code block *for* every value in that variable.

````python
# Using a for loop with a list
my_list = [1, 2, 3, 4, 5]
print("For loop with a list:")
for item in my_list:
    print(item)

# Using a for loop with a tuple
my_tuple = ('a', 'b', 'c', 'd')
print("\nFor loop with a tuple:")
for item in my_tuple:
    print(item)

# Using a for loop with a range
print("\nFor loop with a range:")
for num in range(1, 6):  # Range from 1 to 5
    print(num)

# Using a for loop with a string
my_string = "Python"
print("\nFor loop with a string:")
for char in my_string:
    print(char)
````

## While Loops

While loops allow us to execute a block of code *while* a certain condition is true (this is checked using boolean operators)

````python
# Example of a while loop
counter = 0
print("While loop example:")
while counter < 5:
    print("Counter is:", counter)
    counter += 1  # Increment the counter

# Example of an infinite while loop
print("\nInfinite while loop example:")
while counter:
    print("To infinity, and beyond!")
    counter+=1
    if(counter > 30): # Remove this line and the line below to let it run infinitely
        break  #Warning: if you remove these lines and run it, it may cause your computer to freeze or crash.
````

# Comprehensions
Comprehensions let us to build lists/dictionaries/tuples in one line of code. It's a little confusing at first but with practice you can get the hang of it.

For example, here is standard **for** loop to build a list

split() is a function that just splits strings at the index provided, unless no index is provided (which in this case it isn't). If no index is provided then it splits it exactly down the half.

````python
# List of names
names = ["John Doe", "Jane Smith", "Alice Johnson", "Bob Brown"]

# List to store initials
initials = []

# for loop to extract initials
for name in names:
    first, last = name.split()  # split the name into first and last
    initials.append(first[0] + last[0])  # Add the initials to the list

# print the list of initials
print("List of initials using for loop:", initials)
````

And here is an equivalent list comprehension

````python
names = ["John Doe", "Jane Smith", "Alice Johnson", "Bob Brown"]
# List comprehension to extract initials
initials_listcomp = [name.split()[0][0] + name.split()[1][0] for name in names]
# print the list of initials
print("List of initials using list comprehension:", initials_listcomp)
````

Now if we want to make a dictionary with each name, and the length corresponding to each name we can use a dictionary comprehension.
len(argument) measures the name of whatever string you put in for the argument

````python
names = ["John Doe", "Jane Smith", "Alice Johnson", "Bob Brown"]
# remove the space from the names
names = [name.replace(" ", "") for name in names]
nameslengthdictcomp= {name:len(name) for name in names} # dictionary comprehension
print("Dictionary comprehension with names and their lengths:", nameslengthdictcomp)
````
