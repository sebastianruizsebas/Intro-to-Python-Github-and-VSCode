# Loops
Loops allow us to execute blocks of code multiple times, for example if you want to read through every row of a dataset, or test different values within a function

## For Loops
For loops allow us to execute a block of code a specific number of times.

It iterates through an iterable variable (list, tuple, range, string), executing the code block *for* every value in that variable.

````python
# Using a for loop with a list
mylist = [1, 2, 3, 4, 5]
for item in mylist:
    print(item)
# Using a for loop with a tuple
my_tuple = ('a', 'b', 'c', 'd')
for item in my_tuple:
    print(item)
# Using a for loop with a range
for num in range(1, 6):  # Range from 1 to 5
    print(num)
# Using a for loop with a dictionary
my_dict = {'name': 'Alice', 'age': 25, 'city': 'New York'}
for key, value in my_dict.items():
    print(key, value)
# Using a for loop with a string
my_string = "Python"
for char in my_string:
    print(char)
````

## While Loops

While loops allow us to execute a block of code *while* a certain condition is true (this is checked using boolean operators)

````python
# Example of a while loop
counter = 0
while counter < 5:
    print("Counter is:", counter)
    counter += 1  # Increment the counter

# Example of an infinite while loop
while counter:
    print("To infinity, and beyond!")
    counter+=1
    if(counter > 30): # Remove this line and the line below to let it run infinitely
        break  #Warning: if you remove these lines and run it,
               #it may cause your computer to freeze or crash.
````

## Lets do an experiment similar to the last one
In our experiment, we want to filter out participants from the treatment group who don't meet the minimum criteria

We want our participant to be healthy, not be using drugs, and be younger than 55.

We can use for loops to generate a bunch 'participants' (evil laugh)

````python
import random
ages = []
groups = []
druguses = []
genders = []
# Generate random data for 100 participants
n = 100
for i in range(n):
    ages.append(random.randint(1, 100))
    groups.append(random.choice(["healthy", "pre-existing condition"]))
    druguses.append(random.choice([True, False]))
    genders.append(random.choice(["Male", "Female", "Other"]))
# Generate nested list of participant data
participant_data = []
for i in range(n):
    participant_data.append((i+1, ages[i], groups[i], druguses[i], genders[i]))

#for loop to iterate through the list of participants
accepted_participants = []
rejected_participants = []
for participant in participant_data:
    participant_id, age, group, druguse, gender = participant
    accept_into_treatment = (group == "healthy") and (not druguse) and (age < 55)
    if accept_into_treatment:
        accepted_participants.append(participant)
    else:
        rejected_participants.append(participant)
# Print accepted and rejected participants
print("Accepted Participants:", accepted_participants)

#print a specific participant's data
print(accepted_participants[0])
````

# Comprehensions
Comprehensions let us to build lists/dictionaries/tuples in one line of code. It's a little confusing at first but with practice you can get the hang of it.

Here is code that does the same thing as the previous experiment, except we use list comprehensions instead of for loops

````python
import random

# generate random data for 100 patients using list comprehensions
n = 100
ages = [random.randint(1, 100) for _ in range(n)]
groups = [random.choice(["healthy", "pre-existing condition"]) for _ in range(n)]
druguses = [random.choice([True, False]) for _ in range(n)]
genders = [random.choice(["Male", "Female", "Other"]) for _ in range(n)]

patient_data = [(i + 1, ages[i], groups[i], druguses[i], genders[i]) for i in range(n)]
accepted_participants = [
    patient for patient in patient_data
    if (patient[2] == "healthy") and (patient[3]==False) and (patient[1] < 55)
]
print("First 5 accepted participants:", accepted_participants[0:5])
#the total number of accepted participants
print("Total accepted participants:", len(accepted_participants))
````

Let's write a custom function, expanding on our previous experiment by developing age groups

````python
def group_by_age_range(participants):
    """
    Group participants by age range.
    :param participants: List of participants
    :return: Dictionary with age ranges as keys
    """
    age_groups = {"0-18": [], "19-35": [], "36-55": [], "56+": []}
    for participant in participants:
        age = participant[1]
        if age <= 18:
            age_groups["0-18"].append(participant)
        elif age <= 35:
            age_groups["19-35"].append(participant)
        elif age <= 55:
            age_groups["36-55"].append(participant)
        else:
            age_groups["56+"].append(participant)
    return age_groups
````
